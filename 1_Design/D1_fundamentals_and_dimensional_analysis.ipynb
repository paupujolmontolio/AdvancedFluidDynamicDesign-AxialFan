## Some simple questions

What is the main difference between a **driven turbomachine** (Power-absorving) and a **driving turbomachine** (Power-Generating)? Give some examples of both types. How is that related to the **Euler Equation**?

Answer


A driven turbomachine adds energy to the fluid from it's own driven motion and a driving turbomachine extracts energy from the fluid to generate power. Some driven turbomachines are aircraft propellers, axial fans or pumps... Examples of driving turbomachines are turbines or windmills. The Euler Work Equation has two main versions, one for pumps and one for turbines. Their main difference is the sign for work.

Explain with your words what is the _rothalpy_. Why is that conserved in a rotor?



Answer


Rothalpy is a property of fluid flow in turbomachines related to enthalpy and velocity. It is conserved in a rotor in certain conditions because the work provided by the turbomachine balances the changes in enthalpy and kinetic energy.
